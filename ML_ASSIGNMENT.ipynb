# DATA PREPROCESSING

<div style="background-color: #f0f4f8; padding: 10px; border-left: 5px solid #1a365d; border-radius: 4px;">
  <b style="color: #1a365d; font-size: 16px;">FINDING MISSING VALUES</b>
</div>

```python
df.isnull().sum()
```

```python
missing_percentage = (
                      df.isnull().sum() / len(df)
                     ) * 100
missing_percentage
```

```python
missing_col=["income","savings","monthly_expenses","credit_score"]
```

```python
for col in missing_col:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.isnull().sum()
```

<div style="background-color: #f0f4f8; padding: 10px; border-left: 5px solid #1a365d; border-radius: 4px;">
  <b style="color: #1a365d; font-size: 16px;">FIXING CATEGORICAL INCONSISTENCIES</b>
</div>

```python
df["education"]=df["education"].replace("Bachlors","Bachelors")
```

```python
df["education"].value_counts()
```

```python
df.head(5)
```

```python
df['income_per_dependent'] = (df['income'] /(df['num_dependents'] + 1))
df['income_per_dependent'].head(5)
```

```python
df['signup_date'] = pd.to_datetime(df['signup_date'])
```

```python
reference_date = df['signup_date'].max()
```

```python
df['signup_recency_days'] = (
                               reference_date - df['signup_date']
                            ).dt.days
```

```python
df = df.drop('signup_date', axis=1)
```

<div style="background-color: #f0f4f8; padding: 10px; border-left: 5px solid #1a365d; border-radius: 4px;">
  <b style="color: #1a365d; font-size: 16px;">FIXING OUTLIERS</b>
</div>

```python
outlier_cols = ['income','savings','monthly_expenses','credit_score','loan_amount','debt_to_income','income_per_dependent']
```

```python
for col in outlier_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        df[col] = df[col].clip(lower_bound,upper_bound)
        #clip() basically means "keep values within a specified minimum and maximum."
   
```

```python
df[outlier_cols].boxplot(figsize=(12, 6))
plt.xticks(rotation=45)
plt.title("Boxplots After Outlier Treatment")
plt.show()
```

<div style="background-color: #f0f4f8; padding: 10px; border-left: 5px solid #1a365d; border-radius: 4px;">
  <b style="color: #1a365d; font-size: 16px;">ENCODING VARIABLES</b>
</div>

```python
x = df.drop(columns=["target_default_risk", "customer_id"])
y = df["target_default_risk"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
print("X_train shape:", x_train.shape)
print("X_test shape:", x_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
numeric_col = x_train.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_col = x_train.select_dtypes(include=['object']).columns.tolist()

# target must NEVER be included in X
if 'target_default_risk' in numeric_col:
    numeric_col.remove('target_default_risk')

if 'target_default_risk' in categorical_col:
    categorical_col.remove('target_default_risk')
```

```python
print(categorical_col)
print(numeric_col)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
```

```python
# Numerical preprocessing
numeric_transformer = Pipeline(
                          steps=[
                                  ('imputer', SimpleImputer(strategy='median')),
                                  ('scaler', StandardScaler())
                                ]
                              )

# Categorical preprocessing
categorical_transformer = Pipeline(
                              steps=[
                                      ('imputer', SimpleImputer(strategy='most_frequent')),
                                      ('encoder', OneHotEncoder( handle_unknown='ignore',sparse_output=False))
                                    ]
                                  )

# Combine preprocessing steps
preprocessor = ColumnTransformer(
                                  transformers=[
                                                 ('num', numeric_transformer, numeric_col),
                                                 ('cat', categorical_transformer, categorical_col)
                                               ],  remainder='drop'
                                  )

print("Preprocessor created successfully.")
```

```python
# Fit preprocessing ONLY on training data
x_train_prep = preprocessor.fit_transform(x_train)
print("Processed X_train shape:", x_train_prep.shape)
```

```python
x_test_prep = preprocessor.transform(x_test)
print("Processed X_test shape:", x_test_prep.shape)
```

```python
outlier_cols = ['income','savings','monthly_expenses','credit_score','loan_amount','debt_to_income','income_per_dependent']
```

# HYPERPARAMETER TUNING

```python
from sklearn.model_selection import GridSearchCV
```

```python
rf_param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}
```

```python
rf_grid = GridSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_grid=rf_param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    verbose=1
)

rf_grid.fit(x_train_prep, y_train)
```

```python
rf_best = rf_grid.best_estimator_
rf_tuned_pred = rf_best.predict(x_test_prep)
rf_tuned_acc = accuracy_score(y_test, rf_tuned_pred)
rf_tuned_prec = precision_score(y_test, rf_tuned_pred)
rf_tuned_rec = recall_score(y_test, rf_tuned_pred)
rf_tuned_f1 = f1_score(y_test, rf_tuned_pred)
rf_tuned_cm = confusion_matrix(y_test, rf_tuned_pred)

print("Tuned Random Forest")
print(f"Accuracy:        {rf_tuned_acc:.4f}")
print(f"Precision:       {rf_tuned_prec:.4f}")
print(f"Recall:          {rf_tuned_rec:.4f}")
print(f"F1-Score:        {rf_tuned_f1:.4f}")

print("\nConfusion Matrix:")
print(rf_tuned_cm)
```

```python
xgb_param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [3, 5, 7],
    'learning_rate': [0.05, 0.1],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}
```

```python
xgb_grid = GridSearchCV(
    estimator=XGBClassifier(
        random_state=42,
        eval_metric='logloss'
    ),
    param_grid=xgb_param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    verbose=1
)

xgb_grid.fit(x_train_prep, y_train)
```

```python
xgb_best = xgb_grid.best_estimator_
xgb_tuned_pred = xgb_best.predict(x_test_prep)
xgb_tuned_acc = accuracy_score(y_test, xgb_tuned_pred)
xgb_tuned_prec = precision_score(y_test, xgb_tuned_pred)
xgb_tuned_rec = recall_score(y_test, xgb_tuned_pred)
xgb_tuned_f1 = f1_score(y_test, xgb_tuned_pred)
xgb_tuned_cm = confusion_matrix(y_test, xgb_tuned_pred)

print("Tuned XGBoost")
print(f"Accuracy:        {xgb_tuned_acc:.4f}")
print(f"Precision:       {xgb_tuned_prec:.4f}")
print(f"Recall:          {xgb_tuned_rec:.4f}")
print(f"F1-Score:        {xgb_tuned_f1:.4f}")

print("\nConfusion Matrix:")
print(xgb_tuned_cm)
```
